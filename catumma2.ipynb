```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
df = pd.read_csv("/kaggle/input/datasets/mragpavank/breast-cancer/data.csv")
```

Comparing all classification models and finding a good  classification model based on their metrics

```python
#removing unnecessary sapces from col names
df.columns.str.strip()
```

```python
#removing unnamed columns which are not used 
df = df.drop(columns=["id", "Unnamed: 32"], errors="ignore")
```

```python
X = df.drop(columns=["diagnosis"])
y = df["diagnosis"]
```

```python
X.isnull().sum()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=10,  stratify=y)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_traint = scaler.fit_transform(X_train)
X_testt = scaler.transform(X_test)
```

```python
model = LogisticRegression()
```

```python
model.fit(X_traint,y_train)
```

```python
k=model.predict(X_testt)
k
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test, k)
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(y_test, k)
```

```python
from sklearn.metrics import precision_score
precision_score(y_test, k, pos_label="M")
```

```python
from sklearn.tree import DecisionTreeClassifier
model2=DecisionTreeClassifier(random_state=42)
model2.fit(X_train, y_train)
```

```python
f=model2.predict(X_test)
f
```

```python
from sklearn.metrics import accuracy_score
print("Logistic Regression:",
      accuracy_score(y_test, k))
print("Decision Tree:",
      accuracy_score(y_test, f))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
model3=RandomForestClassifier(n_estimators=10, random_state=42)
model3.fit(X_train, y_train)
print(model3.predict(X_test))
r=model3.predict(X_test)
print("accuracy")
print("RandomForest",accuracy_score(y_test, model3.predict(X_test)))
print("Logistic Regression", accuracy_score(y_test, k))
print("Decision Tree", accuracy_score(y_test, f))
```

```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
model4= SVC(kernel="rbf", C=1.0, random_state=42)
model4.fit(X_traint, y_train)
s=model4.predict(X_testt)
print("accuracy")
print("SVM", accuracy_score(y_test, s))
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
model5 = KNeighborsClassifier(n_neighbors=5)
model5.fit(X_traint, y_train)
k = model5.predict(X_testt)
print("KNN", accuracy_score(y_test, k))
```

```python
print("Logistic Regression:",accuracy_score(y_test, k))
print("Decision Tree:",accuracy_score(y_test, f))
print("Random Forest:",accuracy_score(y_test, r))
print("SVM:",accuracy_score(y_test, s))
print("KNN:",accuracy_score(y_test, k))
```
